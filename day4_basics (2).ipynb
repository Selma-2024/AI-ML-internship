Write a program to check whether a year is a leap year or not.


```python
year = int(input("Enter a year: "))

if (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0):
    print(f"{year} is a leap year.")
else:
    print(f"{year} is not a leap year.")
```

Extract non-repeating (unique) words from a sentence

```python
sent = "i am eating an apple and the apple tastes good while eating"
words = sent.split()

uwords = []
i = 0

while i < len(words):
    if words[i] not in uwords:
        uwords.append(words[i])
    i += 1

print("Unique words:", uwords)
```
