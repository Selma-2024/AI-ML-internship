```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
dataset = pd.read_csv('Social_Network_Ads.csv')
X = dataset.iloc[:, :-1].values
Y = dataset.iloc[:, -1].values
```

```python
X
```

```python
Y
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size = 0.25, random_state = 0)
```

```python
print(X_train)
```

```python
print(X_test)
```

```python
print(Y_train)
```

```python
print(Y_test)
```

# **Feature scaling**

preprocessing technique in machine learning that transforms numerical features to a common scale


min-max
standardisation
robust scaling

```python
from sklearn.preprocessing import StandardScaler
sc=StandardScaler()
X_train=sc.fit_transform(X_train)
X_test=sc.transform(X_test)
```

```python
print(X_train)
```

```python
print(X_test)
```

# **TRAINING**

```python
from sklearn.linear_model import LogisticRegression
classifier=LogisticRegression()
classifier.fit(X_train,Y_train)
```

prediction

```python
print(classifier.predict(sc.transform([[50,40000]])))
```

prediction test results

```python
Y_pred=classifier.predict(X_test)
print(np.concatenate((Y_pred.reshape(len(Y_pred),1),Y_test.reshape(len(Y_test),1)),1))
```

CONFUSION MATRIX

```python
from sklearn.metrics import confusion_matrix, accuracy_score
cm=confusion_matrix(Y_test,Y_pred)
print(cm)
accuracy_score(Y_test,Y_pred)
```

```python
from sklearn.metrics import classification_report, confusion_matrix
print(classification_report(Y_test, Y_pred))
```

```python
from sklearn import metrics
Y_pred_train=classifier.predict(X_train)
print(metrics.accuracy_score(Y_train,Y_pred_train))
print(classification_report(Y_train,Y_pred_train))
```

VISUALISATION - train

```python
from sklearn.inspection import DecisionBoundaryDisplay

DecisionBoundaryDisplay.from_estimator(
    classifier,
    X_train,
    response_method="predict",
    cmap=plt.cm.Spectral,
    alpha=0.8,
    Xlabel=dataset.columns[0],
    Ylabel=dataset.columns[1]
)

plt.scatter(X_train[:, 0], X_train[:, 1], c=Y_train, s=20, edgecolors="k")
plt.title('Logistic Regression (Training set)')
plt.show()
```

VISUALISATION - test

```python
from sklearn.inspection import DecisionBoundaryDisplay

DecisionBoundaryDisplay.from_estimator(
    classifier,
    X_test,
    response_method="predict",
    cmap=plt.cm.Spectral,
    alpha=0.8,
    Xlabel=dataset.columns[0],
    Ylabel=dataset.columns[1]
)

plt.scatter(X_test[:, 0], X_test[:, 1], c=Y_test, s=20, edgecolors="k")
plt.title('Logistic Regression (Training set)')
plt.show()
```
