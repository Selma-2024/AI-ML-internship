```python

students = []

def add_student():
    name = input("Enter student name: ").strip()
    age = int(input("Enter student age: "))

    num_subjects = int(input("Enter number of subjects: "))
    marks = []
    for i in range(num_subjects):
        m = float(input(f"Enter marks for subject {i + 1}: "))
        marks.append(m)

    marks_tuple = tuple(marks)

    student = {
        "name": name,
        "age": age,
        "marks": marks_tuple
    }

    students.append(student)
    print(f"Student '{name}' added successfully!\n")


def calculate_average(marks_tuple):
    """Calculate and return the average of a tuple of marks."""
    if len(marks_tuple) == 0:
        return 0
    return sum(marks_tuple) / len(marks_tuple)


def display_students():
    """Display details of all students using a for loop."""
    if not students:
        print("No student records found.\n")
        return

    print("\n----- Student Records -----")
    for index, student in enumerate(students, start=1):
        avg = calculate_average(student["marks"])
        print(f"{index}. Name: {student['name']}")
        print(f"   Age: {student['age']}")
        print(f"   Marks: {student['marks']}")
        print(f"   Average: {avg:.2f}")
    print("----------------------------\n")


def find_topper():
    """Find and display the student with the highest average marks."""
    if not students:
        print("No student records found.\n")
        return

    topper = None
    highest_avg = -1

    for student in students:
        avg = calculate_average(student["marks"])
        if avg > highest_avg:
            highest_avg = avg
            topper = student

    print("\n----- Topper -----")
    print(f"Name: {topper['name']}")
    print(f"Age: {topper['age']}")
    print(f"Marks: {topper['marks']}")
    print(f"Average: {highest_avg:.2f}")
    print("-------------------\n")


def display_menu():
    """Display the main menu options."""
    print("===== Student Management System =====")
    print("1. Add Student")
    print("2. Display Students")
    print("3. Find Topper")
    print("4. Exit")


def main():
    """Main function containing the while loop to run the menu repeatedly."""
    while True:
        display_menu()
        choice = input("Enter your choice (1-4): ").strip()

        if choice == "1":
            add_student()
        elif choice == "2":
            display_students()
        elif choice == "3":
            find_topper()
        elif choice == "4":
            print("Exiting Student Management System. Goodbye!")
            break
        else:
            print("Invalid choice. Please enter a number between 1 and 4.\n")


if __name__ == "__main__":
    main()
```

# ***Task 2: ATM Machine***

```python
balance = 1000.0
pin = "1234"

def check_balance():
    print(f"\nYour current balance is: {balance:.2f}\n")

def deposit_money():
  print("DEPOSIT MONEY")
  global balance
  amount = float(input("Enter amount to deposit: "))

  if amount <= 0:
      print("Invalid amount. Deposit must be greater than 0.\n")
  else:
      balance += amount
      print(f"Deposit successful! New balance: {balance:.2f}\n")

def withdraw_money():
  print("WITHDRAW MONEY")
  global balance
  amount = float(input("Enter amount to withdraw: "))
  if amount <= 0:
    print("Invalid amount. Withdrawal must be greater than 0.\n")
  elif amount > balance:
    print("Insufficient balance. Transaction failed.\n")
  else:
    balance -= amount
    print(f"Withdrawal successful! New balance: {balance:.2f}\n")


def change_pin():
    print("CHange the pin")
    global pin
    old_pin = input("Enter current PIN: ").strip()

    if old_pin != pin:
        print("Incorrect PIN. Cannot change PIN.\n")
    else:
        new_pin = input("Enter new PIN: ").strip()
        confirm_pin = input("Confirm new PIN: ").strip()

        if new_pin != confirm_pin:
            print("New PIN and confirmation do not match. PIN not changed.\n")
        elif len(new_pin) != 4 or not new_pin.isdigit():
            print("Invalid PIN format. PIN must be a 4-digit number.\n")
        else:
            pin = new_pin
            print("PIN changed successfully!\n")


def verify_pin():
    print("verify")
    entered_pin = input("Enter your PIN: ").strip()

    if entered_pin == pin:
        return True
    else:
        return False




def display_menu():
  print("ATM")
  print("1.check balance")
  print("2. deposit money")
  print("3.withdraw money")
  print("4.change pin")
  print("5.exit")

def main():
  print("Welcome to the ATM!")
  if not verify_pin():
        print("Incorrect PIN. Access denied.")
        return

  print("PIN verified successfully!\n")

  while True:
        display_menu()
        choice = input("Enter your choice (1-4): ").strip()

        if choice == "1":
            check_balance()
        elif choice == "2":
            deposit_money()
        elif choice == "3":
            withdraw_money()
        elif choice == "4":
            change_pin()
        elif choice == "5":
            print("Exiting ATM!")
            break
        else:
            print("Invalid choice. Please enter a number between 1 and 4.\n")

if __name__ == "__main__":
    main()
```
