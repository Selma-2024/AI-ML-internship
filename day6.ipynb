Create a custom module named calculator.py with the following functions:
   a) add()
   b) subtract()
   c) multiply()
   d) divide()

   Then import it into another file and use all the functions.


```python
%%writefile calculator.py
def add(a,b):
  return a+b

def sub(a,b):
  return a-b

def mult(a,b):
  return a*b

def divide(a,b):
  if(b==0):
    print("error")
  else:
    return a/b

```

```python
import calculator as cl
```

```python
a, b = 10, 5

print("Calculator Module Demo")
print("-----------------------")
print(f"Add: {a} + {b} = {cl.add(a, b)}")
print(f"Subtract: {a} - {b} = {cl.sub(a, b)}")
print(f"Multiply: {a} * {b} = {cl.mult(a, b)}")
print(f"Divide: {a} / {b} = {cl.divide(a, b)}")
```

Write a program that imports the random module and simulates rolling a dice (generates a number from 1 to 6).



```python
import random

def roll_dice():
    """Return a random integer between 1 and 6 (inclusive), like a dice roll."""
    return random.randint(1, 6)


if __name__ == "__main__":
    result = roll_dice()
    print(f"You rolled a {result}!")

    print("\nRolling 5 more times:")
    for i in range(1, 6):
        print(f"Roll {i}: {roll_dice()}")
```
